```python
# ============================================================
# Quaternion primitives — XYZW convention
# q = np.array([x, y, z, w])
# i.e. vector part first, scalar part last
# ============================================================

def q_power(q, t):
    """Raise a unit quaternion to the power t. Scales rotation angle by t."""
    q = q / np.linalg.norm(q)
    x, y, z, w = q

    # Decompose to axis-angle
    angle = 2.0 * np.arccos(np.clip(w, -1.0, 1.0))
    s = np.sqrt(1.0 - w*w)
    if s < 1e-9:
        # Near identity — no meaningful axis, return identity
        return np.array([0.0, 0.0, 0.0, 1.0])
    axis = np.array([x, y, z]) / s

    # Scale the angle, rebuild
    new_angle = angle * t
    s_new = np.sin(new_angle / 2.0)
    return np.array([axis[0]*s_new, axis[1]*s_new, axis[2]*s_new,
                     np.cos(new_angle / 2.0)])

def q_normalise(q):
    """Make q a unit quaternion."""
    return q / np.linalg.norm(q)


def q_conjugate(q):
    """Conjugate (= inverse for unit quaternions).
    Flip the sign of the vector part (x, y, z), leave w alone.
    """
    x, y, z, w = q
    return np.array([-x, -y, -z, w])


def q_multiply(q1, q2):
    """Hamilton product q1 ⊗ q2.

    Same math as before, just reordered to put the vector part first
    and the scalar part last in the output array.
    """
    x1, y1, z1, w1 = q1
    x2, y2, z2, w2 = q2
    return np.array([
        w1*x2 + x1*w2 + y1*z2 - z1*y2,   # x
        w1*y2 - x1*z2 + y1*w2 + z1*x2,   # y
        w1*z2 + x1*y2 - y1*x2 + z1*w2,   # z
        w1*w2 - x1*x2 - y1*y2 - z1*z2,   # w
    ])


def q_rotate_vector(q, v):
    """Rotate a 3D vector v by quaternion q via the sandwich product.

    Promotes v to a pure quaternion with w=0, sandwiches, strips w.
    """
    v_quat = np.array([v[0], v[1], v[2], 0.0])    # (x, y, z, w=0)
    q_conj = q_conjugate(q)
    rotated = q_multiply(q_multiply(q, v_quat), q_conj)
    return rotated[:3]                             # drop the scalar (w) part


def q_relative(q_from, q_to):
    """Rotation from q_from to q_to.
       q_rel = q_from^-1 ⊗ q_to
    """
    return q_multiply(q_conjugate(q_from), q_to)


# ============================================================
# Swing-twist decomposition — XYZW
# ============================================================

def twist_about_x(q):
    """Extract the +X-axis twist component of q.
    Zero out y and z, renormalise. w and x stay (now at indices 3 and 0).
    """
    x, y, z, w = q
    denom = np.sqrt(w*w + x*x)
    if denom < 1e-9:
        return np.array([0.0, 0.0, 0.0, 1.0])      # identity in XYZW
    return np.array([x/denom, 0.0, 0.0, w/denom])

def transform_vector_swing_only(v_forearm, q_forearm, q_palm):
    q_f_twist = twist_about_x(q_forearm)
    q_p_twist = twist_about_x(q_palm)
    q_f_swing = q_multiply(q_forearm, q_conjugate(q_f_twist))
    q_p_swing = q_multiply(q_palm,    q_conjugate(q_p_twist))
    q_rel_swing = q_relative(q_p_swing, q_f_swing)
    return q_rotate_vector(q_rel_swing, v_forearm)


def transform_vector_twist_only(v_forearm, q_forearm, q_palm):
    q_f_twist = twist_about_x(q_forearm)
    q_p_twist = twist_about_x(q_palm)
    q_rel_twist = q_relative(q_p_twist, q_f_twist)
    return q_rotate_vector(q_rel_twist, v_forearm)
```
